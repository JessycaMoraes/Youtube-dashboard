## Category and Channel Type

```python
df.category.isnull().sum()
```

This represents 3% of the total.

To fill in the missing values ​​in the category and channel_type columns, we will use the information present in one column to fill in the other, and if both are missing, fill in with ***Unknown***.

```python
def fill_if_missing(row):
    if pd.isna(row['category']) and pd.isna(row['channel_type']):
        return "Unknown", "Unknown"
    elif pd.isna(row['category']):
        return row['channel_type'], row['channel_type']
    elif pd.isna(row['channel_type']):
        return row['category'], row['category']
    else:
        return row['category'], row['channel_type'] # If both have value, we keep the original values
```

```python
# Criar uma máscara booleana para identificar linhas com valores faltantes em 'category' ou 'channel_type'
missing_mask = df['category'].isnull() | df['channel_type'].isnull()
```

```python
# Aplicar a função apenas nas linhas onde a máscara é True
df.loc[missing_mask, ['category', 'channel_type']] = df.loc[missing_mask].apply(
    lambda row: pd.Series(fill_if_missing(row), index=['category', 'channel_type']), axis=1
)

# Você também pode verificar se ainda há valores nulos nessas colunas
print("Number of null values ​​in column 'category':", df['category'].isnull().sum())
print("Number of null values ​​in column 'channel_type':", df['channel_type'].isnull().sum())
```

# Dealing with date

```python
def format_date(year, abbreviated_month, day):
    months = {
        'Jan': 1, 'Feb': 2, 'Mar': 3, 'Apr': 4, 'May': 5, 'Jun': 6,
        'Jul': 7, 'Aug': 8, 'Sep': 9, 'Oct': 10, 'Nov': 11, 'Dec': 12
    }

    if pd.isna(year) or pd.isna(day) or abbreviated_month not in months:
        return None
    else:
        numeric_month = months[abbreviated_month]
        try:
            return datetime(int(year), numeric_month, int(day)).strftime('%Y-%m-%d')
        except ValueError:
            # Error if date is invalid (eg: April 31st)
            return None
```

```python
# Converte ano e dia para numérico, transformando não numéricos em NaN
df['created_year'] = pd.to_numeric(df['created_year'], errors = 'coerce')
df['created_date'] = pd.to_numeric(df['created_date'], errors = 'coerce')
```

```python
df['created_date_formatted'] = df.apply(lambda row: format_date(row['created_year'], row['created_month'], row['created_date']), axis = 1)
print(df[['created_year', 'created_month', 'created_date', 'created_date_formatted']].head())
```
